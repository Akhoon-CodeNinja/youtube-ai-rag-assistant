```python

import pandas as pd
import os

dev_path = "/content/drive/MyDrive/AI_Project_Dataset/dev_final.csv"

dev_df = pd.read_csv(dev_path)

print("Development dataset loaded successfully!")
print("Total records:", len(dev_df))
print("\nColumns:")
print(dev_df.columns.tolist())

print("\nFirst 3 records:")
display(dev_df.head(3))
```

```python

import torch
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

model_path = "/content/drive/MyDrive/AI_Project_Dataset/final_distilbert_model"

tokenizer = AutoTokenizer.from_pretrained(model_path)

model = AutoModelForQuestionAnswering.from_pretrained(model_path)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model.to(device)
model.eval()

print("Model loaded successfully!")
print("Device:", device)
```

```python

import torch

def predict_answer(question, context):
    inputs = tokenizer(
        question,
        context,
        return_tensors="pt",
        truncation=True,
        max_length=384
    )

    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    start_index = torch.argmax(outputs.start_logits)
    end_index = torch.argmax(outputs.end_logits)

    if end_index < start_index:
        return ""

    answer_tokens = inputs["input_ids"][0][start_index:end_index + 1]

    answer = tokenizer.decode(
        answer_tokens,
        skip_special_tokens=True
    )

    return answer


sample = dev_df.iloc[0]

question = sample["question"]
context = sample["context"]
actual_answer = sample["answers"]

predicted_answer = predict_answer(question, context)

print("Question:", question)
print("Actual Answer:", actual_answer)
print("Predicted Answer:", predicted_answer)
```

```python
import evaluate

squad_metric = evaluate.load("squad")
print("SQuAD evaluation metric loaded successfully!")
```

```python

import torch
from tqdm.auto import tqdm

predictions = []
references = []

for i in tqdm(range(len(dev_df))):

    row = dev_df.iloc[i]

    question = row["question"]
    context = row["context"]

    predicted_answer = predict_answer(question, context)

    predictions.append({
        "id": str(i),
        "prediction_text": predicted_answer
    })

    answer = row["answers"]

    references.append({
        "id": str(i),
        "answers": {
            "text": [str(answer)] if not row["is_impossible"] else [],
            "answer_start": [int(row["answer_start"])] if not row["is_impossible"] else []
        }
    })

print("Predictions completed:", len(predictions))
print("References completed:", len(references))
```

```python
import numpy as np
import torch
from tqdm.auto import tqdm

def predict_with_probability(question, context):

    inputs = tokenizer(
        question,
        context,
        max_length=384,
        truncation="only_second",
        return_offsets_mapping=True
    )

    sequence_ids = inputs.sequence_ids()

    offsets = inputs["offset_mapping"]

    model_inputs = {
        key: torch.tensor([value]).to(device)
        for key, value in inputs.items()
        if key in ["input_ids", "attention_mask"]
    }

    with torch.no_grad():
        outputs = model(**model_inputs)

    start_logits = outputs.start_logits[0]
    end_logits = outputs.end_logits[0]

    cls_score = start_logits[0].item() + end_logits[0].item()

    best_score = float("-inf")
    best_answer = ""

    for start in range(len(sequence_ids)):

        if sequence_ids[start] != 1:
            continue

        for end in range(start, min(start + 30, len(sequence_ids))):

            if sequence_ids[end] != 1:
                continue

            score = (
                start_logits[start].item()
                + end_logits[end].item()
            )

            if score > best_score:
                best_score = score

                char_start = offsets[start][0]
                char_end = offsets[end][1]

                best_answer = context[char_start:char_end]

    difference = cls_score - best_score

    no_answer_probability = 1 / (1 + np.exp(-np.clip(difference, -50, 50)))

    return best_answer, float(no_answer_probability)


print("No-answer prediction function created successfully!")
```

```python

squad_v2_predictions = []
squad_v2_references = []

for i in tqdm(range(len(dev_df))):

    row = dev_df.iloc[i]

    predicted_answer, no_answer_probability = predict_with_probability(
        row["question"],
        row["context"]
    )

    squad_v2_predictions.append({
        "id": str(i),
        "prediction_text": predicted_answer,
        "no_answer_probability": no_answer_probability
    })

    answer = row["answers"]

    squad_v2_references.append({
        "id": str(i),
        "answers": {
            "text": [] if row["is_impossible"] else [str(answer)],
            "answer_start": [] if row["is_impossible"] else [int(row["answer_start"])]
        }
    })

print("SQuAD 2.0 predictions:", len(squad_v2_predictions))
print("SQuAD 2.0 references:", len(squad_v2_references))
print("Prediction generation completed!")
```
